```python
# calculating range using numpy

import numpy as np


data = np.array([12, 18, 25, 30, 42])
range_value = np.max(data) - np.min(data)

print(f"Range : {range_value}")
```

```python
#example
data = np.array([10, 20, 30, 40, 50])

range_value = np.max(data) - np.min(data)

print(f"Range : {range_value}")
```

```python
# calculating range using pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [45, 50, 55, 60, 70, 80]
})

range_value = df["marks"].max() - df["marks"].min()

print(f"Range : {range_value}")
```
