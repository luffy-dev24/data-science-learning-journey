```python
# calculating quartiles using numpy

import numpy as np

data = np.array([12, 18, 25, 31, 40, 47, 55, 63, 72])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1 ,"25% of values in dataset are at or below ", q1)
print("Q2:", q2 ,"50% of values in dataset are at or below ", q2)
print("Q3:", q3, "75% of values in dataset are at or below ", q3)

#another way of find Q2
q2 = np.median(data)
print(f"Q2: {q2}")

```

```python
data = np.array([
    15, 20, 25, 30, 35,
    40, 45, 50, 55, 60
])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)


print("Q1:", q1 ,", 25% of values in dataset are at or below ", q1)
print("Q2:", q2 ,", 50% of values in dataset are at or below ", q2)
print("Q3:", q3, ", 75% of values in dataset are at or below ", q3)
```

```python
# Quartiles with Pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [12, 18, 25, 31, 40, 47, 55, 63, 72]
})

q1 = df["marks"].quantile(0.25)
q2 = df["marks"].quantile(0.50)
q3 = df["marks"].quantile(0.75)

print("Q1:", q1 ,", 25% of values in dataset are at or below ", q1)
print("Q2:", q2 ,", 50% of values in dataset are at or below ", q2)
print("Q3:", q3, ", 75% of values in dataset are at or below ", q3)
```

```python
# we can calculate all at once
quartile = df["marks"].quantile([0.25,0.50,0.75])
print(quartile)
```
