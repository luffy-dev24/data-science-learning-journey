```python
import numpy as np
```

```python
# calculating std using numpy

data = np.array([10, 12, 14, 16, 18])
std = np.std(data)
print(f"Standard Deviation : {std}")
```

```python
#eaxmple

data = np.array([5, 7, 9, 11, 13])
std = np.std(data)
print(f"Standard Deviation : {std}")
```

```python
#calculating std using pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [10, 12, 14, 16, 18]
})
sd = df["marks"].std()

print("Standard Deviation:", sd) #just like variance in pandas , pandas calculate sample std by default

#to calculate the std of population we must add "ddof=0" 

pop_std = df["marks"].std(ddof=0)
print("Population Standard Deviation:", pop_std)
```

```python
#example 

df = pd.DataFrame({
    "marks": [5, 7, 9, 11, 13]
})
std = df["marks"].std(ddof=0)

print("Standard Deviation:", std)
```

```python
# real world example

"""
problem : Imagine two factories manufacture the same type of component. The target weight is 100 grams.

Factory A : 98, 99, 100, 101, 102
Factory B : 90, 95, 100, 105, 110

both have mean  = 100

calculate standard deviation 
"""

factory_A = np.array([98, 99, 100, 101, 102])
factory_B = np.array([90, 95, 100, 105, 110])

std_factory_A = np.std(factory_A)
std_factory_B = np.std(factory_B)

print(f"Standard Deviation of Factory A : {std_factory_A}")
print(f"Standard Deviation of Factory B : {std_factory_B}")

if std_factory_A < std_factory_B:
    print(f"Factory A has smaller Standard Deviation")
else:
    print(f"Factory B has smaller Standard Deviation")
```
