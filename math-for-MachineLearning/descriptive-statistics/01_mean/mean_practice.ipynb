```python
# mean manual calculation using python

data = [10, 20, 30, 40, 50]
total = sum(data)
n = len(data)

mean = total / n

print(f"Mean : {mean} ") #mean is number that represents the whole group
```

```python
#another example

data = [15, 25, 30, 40, 60, 80]
total = sum(data)
n = len(data)

mean = total / n

print(f"Mean : {mean} ") 
```

```python
# calculating mean using numpy

import numpy as np

data = np.array([10, 20, 30, 40, 50])
mean = np.mean(data)

print(f"Mean : {mean}")
```

```python
#calculating mean using pandas 

import pandas as pd

df = pd.DataFrame({
    "marks": [45, 50, 55, 60, 70, 80]
})

print(df["marks"].mean())
```

```python
#outliers

data = np.array([20, 22, 24, 26, 28])
mean = np.mean(data)
print(f"Mean : {mean}")

data = np.array([20, 22, 24, 26, 100])
mean = np.mean(data)
print(f"Mean : {mean}")

#so extreme values will pull the towards themself , so mean will affected by outliers
```

```python
# Real-World Problem
dataset = [30, 32, 35, 36, 38, 40, 42] #monthly salaries in thousands of rupees
data = np.array(dataset)

mean = np.mean(data)
print(f"Mean : {mean}")

#now adding another person salary 500 (might be CEO)
dataset = [30, 32, 35, 36, 38, 40, 42,500]
data = np.array(dataset)

mean = np.mean(data)
print(f"Mean : {mean}")

# the original mean : 36.142857142857146
# the new mean  : 94.125
#there is one extreme values (500) pulls the mean 
# 500 is may be an outlier , iam saying may be because 500 might be CEO's salary 
```

```python
# ML-Style Challenge 

prices = np.array([
    45, 48, 50, 52, 55,
    58, 60, 62, 65, 500
])

mean = np.mean(prices)
print(f"Mean : {mean}")
print(f"Median : {np.median(prices)}")


#here there is an extreme value , so mean will affected by by extreme values , so if we want one 
#number that represents the whole group and center of the data , i would choose Median , because median 
#will not affected by extreme values , it just represents the center of the data
```
