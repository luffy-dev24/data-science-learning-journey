```python
# calculating mode using numpy
from scipy import stats

data = [3, 5, 7, 5, 9, 5, 11]

mode_result = stats.mode(data, keepdims=True)

print(mode_result)
```

```python
#calculating mode using pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [45, 50, 50, 60, 65, 50, 70]
})

df["marks"].mode()
```

```python
languages = pd.DataFrame([
    "Python",
    "Java",
    "Python",
    "JavaScript",
    "Python",
    "C++",
    "Java" ])
languages.mode()
```

```python
df = pd.DataFrame({
    "payment_method": [
        
        "UPI",
        "Cash",
        "UPI",
        "Card",
        "UPI",
        "Card",
        "UPI",
        "Cash"
    ]
})
print(df.mode())
```

```python
data = [2, 2, 4, 4, 6, 8, 10]
mode = stats.mode(data , keepdims=True)
print(mode)
# here the number 2 is appeared 2 times compare to ather values so mode 2
```
