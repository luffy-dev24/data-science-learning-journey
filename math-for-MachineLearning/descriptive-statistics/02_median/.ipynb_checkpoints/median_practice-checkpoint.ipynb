```python
#calculating median using numpy

import numpy as np

data = np.array([18, 7, 9, 25, 12])
median = np.median(data)

print(f"Median : {median}")
```

```python
# calculating median using pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [45, 52, 60, 65, 70, 78, 85]
})

median  = df["marks"].median()
print(f"Median  : {median}")

mean = df["marks"].mean()
print(f"Mean : {mean}")

# in this example both mean and median are same
```

```python
# Outlier Practice

data = np.array([20, 22, 24, 26, 28])

print("Before Adding Outlier")
print(f" Mean :  {np.mean(data)}" )
print(f" Median : {np.median(data)}")

#in this example both mean and median are same 
#so lets add one extreme value

data = np.array([20, 22, 24, 26, 100])

print("After Adding Outlier")
print(f" Mean :  {np.mean(data)}" )
print(f" Median : {np.median(data)}")

#so by observing the outputs , mean is most afftected by outlier
#so in case of outliers , we can choose median than mean
```

```python
#Real-World Problem

data = np.array([25, 28, 30, 32, 35, 38, 40]) #salaries in thousands
print(f" Mean :  {np.mean(data)}" )
print(f" Median : {np.median(data)}")

data = np.array([25, 28, 30, 32, 35, 38, 40, 500]) #including CEO's salary
print(f" Mean :  {np.mean(data)}" )
print(f" Median : {np.median(data)}")

#so here mean is affected by extreme values , si in this case we can choose median to represent the typical employee salary
```

```python
prices = np.array([
    45, 48, 50, 52, 55,
    58, 60, 62, 65, 500
])

print(f" Mean :  {np.mean(prices)}" )
print(f" Median : {np.median(prices)}")

# so here also the extreme values (500) pulls the mean , so i would choose median to represent the typical house price
```
