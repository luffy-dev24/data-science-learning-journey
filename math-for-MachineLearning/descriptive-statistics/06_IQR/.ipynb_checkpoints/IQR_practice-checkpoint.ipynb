```python
# calculating IQR using numpy 
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
#example

data = np.array([5, 8, 12, 15, 20, 25, 30, 40, 50])

q1 = np.percentile(data , 25)
q3 = np.percentile(data , 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
# calculating IQR using pandas
import pandas as pd

df = pd.DataFrame({
    "marks": [10, 20, 30, 40, 50, 60, 70]
})

q1 = df["marks"].quantile(0.25)
q3 = df["marks"].quantile(0.75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```
