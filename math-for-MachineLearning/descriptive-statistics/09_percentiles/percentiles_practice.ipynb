```python
# calculating percentile using numpy and pandas
import numpy as np
import pandas as pd
```

```python
#example 1 :

import numpy as np

data = np.array([5, 10, 15, 20, 25, 30, 35, 40, 45])

per_25 = np.percentile(data,25)
print(f"25th Percentile : {per_25}")

per_50 = np.percentile(data,50)
print(f"50th Percentile : {per_50}")

per_75 = np.percentile(data,75)
print(f"75th Percentile : {per_75}")

per_90 = np.percentile(data,90)
print(f"90th Percentile : {per_90}")
```

```python
#example 2 : using pandas

df = pd.DataFrame({
    "marks": [5, 10, 15, 20, 25, 30, 35, 40, 45]
})

per_25 = df["marks"].quantile(0.25)
print(f"25th Percentile : {per_25}")

per_50 = df["marks"].quantile(0.50)
print(f"50th Percentile : {per_50}")

per_75 = df["marks"].quantile(0.75)
print(f"75th Percentile : {per_75}")

per_90 = df["marks"].quantile(0.90)
print(f"90th Percentile : {per_90}")
```
