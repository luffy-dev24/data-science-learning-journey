```python
# calculating variance using numpy 

import numpy as np

data = np.array([5, 7, 9, 11, 13])
variance = np.var(data)

print("Variance:", variance)
```

```python
#example 

data = np.array([4, 6, 8])
variance = np.var(data)

print(f"Variance : {variance}")
```

```python
#example 

data = np.array([48, 49, 50, 51, 52])
variance = np.var(data)

print(f"Variance : {variance}")
```

```python
# calculating variance using pandas

import pandas as pd

df = pd.DataFrame({
    "marks": [48, 49, 50, 51, 52]
})

variance = df["marks"].var()

print(f"Variance : {variance}") 

#NumPy → population variance by default
#Pandas → sample variance by default
#We can make Pandas calculate population variance using:
#df["marks"].var(ddof=0)

variance = df["marks"].var(ddof=0)

print(f"Variance : {variance}") #population variance
```

```python
#example  

df = pd.Series([5, 7, 9, 11, 13])

variance = df.var(ddof=0)

print(f"Variance : {variance}") 
```

```python
# comparing two datasets

A = np.array([48, 49, 50, 51, 52])
B = np.array([40, 45, 50, 55, 60])

variance1 = np.var(A)
variance2 = np.var(B)

print(f"Variance of dataset A : {variance1}") #variance is 2.0   
print(f"Variance of dataset B : {variance2}") # variance is 50.0

#so the dataset B values are more spread out from the mean than compare to dataset A
```

```python
# real world examples

# problem : 
"""
Imagine two machines produce parts.

Machine_A = np.array([99, 100, 101, 100, 100])
Machine_B = np.array([90, 110, 100, 95, 105])

Calculate the variance of both using NumPy.

Then answer:

Which machine produces more consistent parts, and why?
"""

Machine_A = np.array([99, 100, 101, 100, 100])
Machine_B = np.array([90, 110, 100, 95, 105])

variance_of_machine_A = np.var(Machine_A)
variance_of_machine_B = np.var(Machine_B)

print(f"Variance of Machine_A : {variance_of_machine_A}")   
print(f"Variance of Machine_B : {variance_of_machine_B}") 

if variance_of_machine_A < variance_of_machine_B :
    print(f"machine A produces more consistent parts")
else:
    print(f"machine B produces more consistent parts")
```

```python
# real world problem

df = pd.DataFrame({
    "student_A": [48, 49, 50, 51, 52],
    "student_B": [40, 45, 50, 55, 60]
})

variance_A = df["student_A"].var(ddof=0)
variance_B = df["student_B"].var(ddof=0)

print(f"Variance of student_A : {variance_A}")   
print(f"Variance of student_B : {variance_B}")

if variance_A > variance_B:
    print(f"student A marks are more spread out")
else:
    print(f"student B marks are more spread out")
```
