```python
#calculating Z-score 
import numpy as np
import pandas as pd
```

```python
#example of z-score using numpy

salaries = np.array([40, 45, 50, 55, 90])

mean = np.mean(salaries)
std = np.std(salaries)
z_scores = (salaries - mean) / std

print("Mean:", mean)
print("Std:", round(std, 2))
print("Z-scores:", np.round(z_scores, 2))
```

```python
#example of z-score using pandas

df = pd.DataFrame({
    'employee': ['A', 'B', 'C', 'D', 'E'],
    'salary': [40, 45, 50, 55, 90]
})

mean = df["salary"].mean()
std = df["salary"].std(ddof=0)
df["deviations"] = df["salary"] - mean
df["z_scores"] = df["deviations"] / std

print(df["z_scores"])
```

```python
#example 

data = np.array([12, 15, 18, 20, 25])
mean = np.mean(data)
std = np.std(data)
z_scores = (data - mean) / std

print(f"Mean : {mean}")
print(f"std : {std}")
print("z_scores : ",z_scores)
```

```python
#example

df = pd.DataFrame({"City": ["Delhi","Mumbai","London","Moscow","Dubai","Tokyo"],
                  "Temperature":[38,32,18,5,42,25]},)

df["Avearge"] = df["Temperature"].mean()
df["std"] = df["Temperature"].std(ddof=0)

df["deviations"] = df["Temperature"] - df["Avearge"]
df["z_score"] = df["deviations"]/df["std"]
```

```python
#example

data = np.array([5, 7, 7, 9, 11])
mean = np.mean(data)
std_with_ddof_0 = np.std(data , ddof=0)
std_with_ddof_1 = np.std(data , ddof=1)

print(f"std with ddof=0  : {std_with_ddof_0}")
print(f"std with ddof=1  : {std_with_ddof_1}")



zscore_with_std_0 = (data - mean)/std_with_ddof_0
zscore_with_std_1 = (data - mean)/std_with_ddof_1

print(f"z-scores with ddof=0  : {zscore_with_std_0}")
print(f"z-scores with ddof=1  : {zscore_with_std_1}")
```
