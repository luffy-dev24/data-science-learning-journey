```python
import matplotlib.pyplot as plt
```

```python
#problems 1
days = [1, 2, 3, 4, 5]

study = [2, 3, 4, 3, 5]
sleep = [8, 7, 8, 6, 7]
play = [3, 4, 2, 5, 3]

labels= ["days","study","sleep","play"]

plt.figure(figsize=(8,5))
plt.stackplot(days,study,sleep,play,labels=labels,alpha=0.8)
plt.legend(loc="upper left")
plt.title('Weekly Activity Tracker')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.grid(True)
plt.show()
```

```python
#Problem 6: Sales by product
Months= [1, 2, 3, 4, 5, 6]

Phones= [10, 12, 14, 13, 15, 18]
Laptops= [8, 9, 7, 10, 11, 12]
Tablets= [4, 5, 6, 5, 7, 8]
labels = ["Phones","Laptops","Tablets"]
colors = ["#4C78A8", "#F58518", "#54A24B"]

plt.stackplot(Months,Phones,Laptops,Tablets,labels=labels,alpha=0.5,colors=colors)
plt.legend(loc="upper left")
plt.xlabel("Months")
plt.ylabel("sales")
plt.title("Sales of Products")
plt.show()
```
