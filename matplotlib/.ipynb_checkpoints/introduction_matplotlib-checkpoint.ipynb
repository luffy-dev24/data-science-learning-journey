```python
import matplotlib.pyplot as plt
```

```python
#matplotlib.pyplot is a module in matplotlib used to visualize the data using 2D lines and markers 
```

```python
plt.plot([1,2,3],[4,5,6])
plt.show() #function to display the plot
```

```python
#real data example

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
#we can style the plot by adding labels so we can understand plot better
plt.plot(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
#multiple lines in one plot

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years , kohli , label="Virat Kohli")
plt.plot(years , sehwag , label="Virender Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend()
```

```python
#using format string
plt.plot(years , kohli , "ro--" , label="Virat Kohli")
plt.plot(years, sehwag , "b^--" , label="Virendra Sehwag")
plt.legend()
```

```python
#using color and line style arguments
plt.plot(years, kohli , color="red" , linestyle="--" , label="Virat Kohli")
plt.plot(years , sehwag , color="blue" , linestyle="--" , label="Virendra Sehwag")
plt.legend()
```

```python
#line width
plt.plot(years, kohli, color="red", linestyle="--", linewidth=3, label="Virat Kohli")
plt.plot(years, sehwag, color="blue", linestyle="--", linewidth=2 , label="Virendra Sehwag")
plt.legend()
```

```python
#layout
plt.plot(years, kohli, color="red", linestyle="--", linewidth=3, label="Virat Kohli")
plt.plot(years, sehwag, color="blue", linestyle="--", linewidth=2 , label="Virendra Sehwag")
plt.legend()
plt.grid(True)
plt.tight_layout()
```

```python
#available styles 
print(plt.style.available)
```

```python
plt.plot(years, kohli, color="red", linestyle="--", linewidth=3, label="Virat Kohli")
plt.plot(years, sehwag, color="blue", linestyle="--", linewidth=2 , label="Virendra Sehwag")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.style.use('petroff10')
```

```python
#XKCD style
with plt.xkcd():
    plt.plot(years, kohli, label="Kohli")
    plt.plot(years, sehwag, label="Sehwag")
    plt.title("Epic Battle of the Batsmen")
    plt.legend()
    plt.show()
```
