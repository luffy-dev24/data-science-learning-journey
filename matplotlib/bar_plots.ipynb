```python
import matplotlib.pyplot as plt
```

```python
categories = ["Grocery", "Electronics", "Clothing", "Toys", "Furniture"]
sales = [45, 80, 60, 25, 35]  # ₹ lakhs

plt.bar(categories,sales)
plt.xlabel("products")
plt.ylabel("sales is lakhs")


for i in range(len(sales)):
    plt.text(i,sales[i]+0.6,str(sales[i]),ha="center")

    
plt.show()
```
