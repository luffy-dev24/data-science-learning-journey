```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 6]

plt.scatter(x, y,s=100,marker='^',c=y, cmap='viridis')
plt.xlabel("X values")
plt.ylabel("Y values")
plt.colorbar()
plt.title("My First Scatter Plot")
plt.show()
```

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [35, 40, 50, 55, 65, 70, 80, 90]

plt.scatter(hours,marks,c=marks,cmap="viridis",s=40,marker="^")
for i in range(len(marks)):
    plt.annotate(f"s{i+1}",(hours[i]+0.2,marks[i]))

plt.xlabel("study hours")
plt.ylabel("marks")
plt.title("Relation between study hours and marks")
plt.grid(True)
plt.show()
```
