```python
import matplotlib.pyplot as plt
```

```python
#daily screen time breakdown

days = [1, 2, 3, 4, 5, 6, 7]
social_media = [2, 3, 1.5, 2.5, 3, 4, 3.5]
gaming = [1, 1, 2, 1.5, 1, 0.5, 2]
work = [4, 5, 4.5, 5, 6, 2, 1]
labels=["social_media","gaming","work"]
colors = ["#C274E8","#74AAE8","#74E8A0"]

plt.stackplot(days,social_media,gaming,work,labels=labels,colors=colors)

plt.title("daily screen time breakdown")
plt.xlabel("days")
plt.ylabel("hours")
plt.legend(loc="upper right")
plt.show()
```

```python
#Company Revenue by Product Line

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
product_a = [20, 25, 22, 30, 35, 40]
product_b = [15, 18, 20, 19, 22, 25]
product_c = [10, 8, 12, 15, 14, 18]
labels=["Product A", "Product B","Product C"]
plt.stackplot(months,product_a,product_b,product_c,labels=labels,colors=colors)
plt.grid(True)
plt.legend(loc="upper left",bbox_to_anchor=(1, 1))
plt.show()
```
