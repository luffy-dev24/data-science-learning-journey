```python
import matplotlib.pyplot as plt
```

```python
x = ["Sleep", "Work", "Study", "Leisure"]
hours = [8,9,3,4]

explode=[0,0.1,0,0]
plt.pie(hours,labels=x,autopct="%1.1f%%",explode=explode)
plt.show()
```

```python
#Mobile OS Market Share
mobiles = ["Android","iOS","Others"]
shares = [70,28,2]
explode=[0.1,0,0,]
plt.pie(shares,labels=mobiles , autopct="%1.1f%%",explode=explode,wedgeprops={"edgecolor":"black","linewidth":1,"linestyle":"--"})
plt.show()
```

```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels , autopct="%1.1f%%",wedgeprops={"edgecolor":"black","linewidth":1,"linestyle":"--"})
plt.show()
```
