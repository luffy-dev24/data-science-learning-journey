```python
import matplotlib.pyplot as plt
```

```python
plt.style.available
plt.style.use("ggplot")
```

```python
# basic pie chart

labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
plt.pie(runs , labels=labels)
plt.tight_layout()
```

```python
#custom colors
colors = ["#F58727","#F2E81D","#1DF2AF","#AB1DF2"]
plt.pie(runs,labels=labels,colors=colors)
plt.show()
```

```python
# Add Edges and Style Slices with wedgeprops
plt.pie(runs,labels=labels,colors=colors, wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'})
plt.show()
```

```python
#Highlight a Slice Using explode
explode = [0.1,0,0,0]
plt.pie(runs,labels=labels,colors=colors,explode=explode)
plt.show()
```

```python
#Add Shadows for a 3D 
explode =[0.1,0,0,0]
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    explode=explode,
    shadow=True
)
plt.show()
```

```python
#Start Angle
plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True,
    startangle=20
)
plt.show()
```

```python
#Show Percentages with autopct
plt.pie(
    runs,
    labels=labels,
    autopct='%1.2f%%',
    startangle=140,
    colors=colors
)
plt.title("Career Run Share")
plt.show()
```
