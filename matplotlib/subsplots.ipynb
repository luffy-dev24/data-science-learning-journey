```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y1 = [i * 2 for i in x]
y2 = [i ** 2 for i in x]
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("double of x")

plt.subplot(1,2,2)
plt.plot(x,y2)
plt.title("square of x")
```

```python

fruits = ['Apples', 'Bananas', 'Mangoes', 'Grapes', 'Oranges']
sales = [23, 45, 12, 38, 19]

fig,axs = plt.subplots(1,2,figsize=(14, 4))

axs[0].bar(fruits,sales,color="skyblue")
axs[0].set_xlabel("fruits")
axs[0].set_ylabel("Sales")
axs[0].set_title("comparision of fruits sales")

axs[1].barh(fruits,sales,color="orange")
    
```
