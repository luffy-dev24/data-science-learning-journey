```python
import matplotlib.pyplot as plt
```

```python
fruits = ['Apples', 'Bananas', 'Mangoes', 'Grapes', 'Oranges']
sales = [23, 45, 12, 38, 19]
plt.bar(fruits,sales,color="skyblue")
plt.ylabel("sales")
plt.xlabel("fruits")
plt.title("comparision of fruits sales")

for i in range(len(fruits)):
    plt.text(i , sales[i]+0.5 , str(sales[i]) , ha="center")
```

```python
plt.barh(fruits,sales,color="orange")
plt.ylabel("fruits")
plt.xlabel("sales of fruits")

for i in range(len(fruits)):
    plt.text(sales[i]+0.5 , i , str(sales[i]))
```

```python
# students marks
students = ['Riya', 'Aman', 'Kabir', 'Sneha', 'Omar']
marks = [78, 65, 90, 55, 82]

plt.bar(students , marks , color = ["red" if val<60 else "green" for val in marks])

plt.xlabel("students")
plt.ylabel("Marks")
plt.title("Students Marks")

def results(val):
    if val<60:
        return "fail"
    elif val>60:
        return "pass"
    
for i in range(len(students)):
    plt.text(i,marks[i]+0.9 ,results(marks[i]), ha="center")
```

```python
# Movie Ratings (Horizontal)

movies = ['Inception', 'Interstellar', 'Tenet', 'Dunkirk', 'Oppenheimer']
ratings = [8.8, 8.6, 7.3, 7.8, 8.4]

plt.barh(movies,ratings,color="orange")
plt.xlabel("ratings")
plt.ylabel("movies")
plt.title("Movies Ratings")

for i in range(len(movies)):
    plt.text(ratings[i],i,str(ratings[i]) )
```

```python
# Grouped Bar Chart: Weekly Temperature
import numpy as np
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
city_a = [30, 32, 29, 31, 33]
city_b = [25, 27, 26, 28, 24]


plt.xlabel("days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature")
width=0.2

x = np.arange(len(days))

plt.bar(x-width, city_a , width=width+0.1 ,  color="red", label="City A" )
plt.bar(x+width, city_b ,width=width+0.1, color="blue" , label="City B")
plt.xticks(x,days)

plt.legend(fontsize=6)


for i in range(len(days)):
    plt.text(i-width , city_a[i]+0.5 , str(city_a[i]) , ha="center")


for i in range(len(days)):
    plt.text(i+width , city_b[i]+0.5 , str(city_b[i]) , ha="center")
```

```python
import numpy as np

categories = ['Food', 'Rent', 'Transport', 'Entertainment']

jan = [200, 800, 150, 100]
feb = [180, 600, 160, 120]

x = np.arange(len(categories))
width=0.2

plt.bar(x-width, jan, width=width+0.1 ,label='Jan', color='skyblue')
plt.bar(x+width, feb, width=width+0.1 ,label='Feb', color='orange')

plt.ylabel("Amount spent")
plt.title("Monthly Budget Breakdown")
plt.legend()
plt.xticks(x,categories)

for i in range(len(categories)):
    plt.text(i-width , jan[i] , str(jan[i]) , ha="center")
for i in range(len(categories)):
    plt.text(i+width , feb[i] , str(feb[i]) , ha="center")

```
