```python
import matplotlib.pyplot as plt
```

```python
#Marks of 15 students

marks = [45, 52, 58, 61, 63, 65, 67, 70, 72, 74, 78, 81, 85, 90, 95]
plt.hist(marks,bins=5,edgecolor="black")
plt.title("marks of students")
plt.xlabel("marks")
plt.ylabel("number of students")
plt.show()
```

```python
#customs bins
plt.hist(marks,bins=[40, 50, 60, 70, 80, 90, 100],edgecolor="black")
plt.title("marks of students")
plt.xlabel("marks")
plt.ylabel("number of students")
plt.show()
```

```python
Ages= [18, 22, 25, 25, 27, 29, 30, 31, 33, 35, 35, 38, 40, 42, 45, 50, 55, 60]
plt.hist(Ages,bins=6,edgecolor="black")
plt.show()
```

```python
#customs bins
Ages= [18, 22, 25, 25, 27, 29, 30, 31, 33, 35, 35, 38, 40, 42, 45, 50, 55, 60]
plt.hist(Ages,bins=[10,20,30,40,50,60,70,80,90],edgecolor="black")
plt.title("Age distribution")
plt.xlabel("Ages")
plt.ylabel("Number of people")
plt.show()
```
