```python
import pandas as pd
```

```python
df = pd.read_csv("regional_sales_raw.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
df["store"] = df["store"].str.strip().str.lower()
```

```python
df["category"] = df["category"].str.strip().str.lower()
```

```python
df["payment_method"] = df["payment_method"].str.strip().str.lower()
```

```python
df = df.drop_duplicates(subset=df.columns.drop("txn_id"))
```

```python
df["rating"] = df["rating"].fillna(0)
```

```python
df["customer_age"] = df["customer_age"].fillna(df["customer_age"].mean())
```

```python
df=df.dropna()
```

```python
df["txn_date"] = pd.to_datetime(df["txn_date"] , format="mixed", errors="coerce")
```

```python
df
```

```python
df = df[(df["unit_price"]>0) & (df["customer_age"].between(0,100))]
```
