```python
names_list = [ 
    ["luffy",19],
    ["zoro",22],
    ["sanji",22],
    ["nami",20],
]
#pd.DataFrame(names_list)
pd.DataFrame(names_list , columns=["Name","Age"]) #DataFrame from python list
```
