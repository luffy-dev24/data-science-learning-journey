```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Ravi', 'Anu', 'Kiran'],
    'math': [80, 90, 75],
    'science': [70, 85, 95]
})
print(df)
```

```python
long_df = df.melt(id_vars = "name" ,var_name="subject",value_name="marks")
print(long_df)
```

```python
wide_df = long_df.pivot(index="name",columns="subject",values="marks")
print(wide_df)
```

```python
sales = pd.DataFrame({
    'city': ['Hyd', 'Hyd', 'Delhi', 'Delhi'],
    'month': ['Jan', 'Jan', 'Jan', 'Feb'],
    'amount': [100, 200, 150, 300]
})

# sales.pivot(index='city', columns='month', values='amount')  # ValueError: Hyd-Jan repeats


dff = sales.pivot_table(index='city', columns='month', values='amount', aggfunc='sum')
print(dff)
```

```python
df1 = pd.DataFrame({
    'city': ['Hyderabad', 'Pune'],
    '2022': [10, 7],
    '2023': [11, 8]
})
long_df1 = df1.melt(id_vars="city",var_name="years",value_name="population")
print(long_df1)
wide_df1 = long_df1.pivot_table(index="city",columns="years",values="population",aggfunc="sum")
print(wide_df1)
```

```python
df2 = pd.DataFrame({
    'product': ['A', 'B'],
    'q1': [100, 200],
    'q2': [150, 250],
    'q3': [120, 220]
})
long_df2 = df2.melt(id_vars="product",var_name="quarter",value_name="revenue")
print(long_df2)
```
