```python
import pandas as pd
```

```python
df = pd.read_csv("sales_messy.csv").copy()
```

```python
df = df.dropna(subset="customer").copy()
df
```

```python
df["product"] = df["product"].str.lower()
```

```python
df
```

```python
df = df[ (df["quantity"]>0 )& (df["quantity"]<30)].copy()
```

```python
def removestr(x):
    x = str(x)
    if x.startswith("$"):
        x = x.strip("$")
    return float(x )   
```

```python
df["price"] = df["price"].apply(removestr)
```

```python
df
```

```python
df = df.dropna(subset="price").copy()
```

```python
df
```

```python
df["order_date"] = pd.to_datetime(df["order_date"] , format="%d/%m/%Y" , errors="coerce")
```

```python
df = df.dropna(subset="order_date").copy()
```

```python
df
```

```python
df["status"] = df["status"].str.lower()
```

```python
df["region"] = df["region"].str.lower()
```

```python
df
```

```python
df.duplicated(subset=[i for i in df.columns[1:]])
```

```python
df = df.drop_duplicates(subset=[i for i in df.columns[1:]]).copy()
```
