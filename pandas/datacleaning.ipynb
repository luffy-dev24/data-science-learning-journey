```python
import pandas as pd

df = pd.DataFrame({"date": ["2024-01-15", "2024-02-20", "2024-03-05"]})
print(df.dtypes)
```

```python
df["date"] = pd.to_datetime(df["date"])
print(df.dtypes)
```

```python
dff = pd.DataFrame({"birthday" : ["2000-05-10", "1998-11-23", "2001-07-01"]})
print(dff)
print(dff.dtypes)
```

```python
dff["birthday"] = pd.to_datetime(dff["birthday"])
print(dff)
print(dff.dtypes)
```

```python
df = pd.DataFrame({"date": ["15/01/2024", "20/02/2024", "05/03/2024"]})
df["date"] = pd.to_datetime(df["date"], format="%d/%m/%Y")
print(df)
```

```python
df = pd.DataFrame({"event_day" : ["March 5, 2024", "April 10, 2024", "July 4, 2024"]})
print(df)
df["event_day"] = pd.to_datetime(df["event_day"] , format="%B %d, %Y")
print(df)
```

```python
df = pd.DataFrame({"log_time":["2024-01-15 14:30:00", "2024-02-20 09:05:00", "2024-03-05 23:59:00"]})
print(df)
df["log_time"] = pd.to_datetime(df["log_time"] , format="%Y-%m-%d %H:%M:%S")
print(df)
```

```python
df = pd.DataFrame({"date": ["01/02/2024", "03/04/2024"]})

print(pd.to_datetime(df["date"]))                  # assumes MM/DD -> Jan 2, Mar 4
print(pd.to_datetime(df["date"], dayfirst=True))    # assumes DD/MM -> Feb 1, Apr 3
```

```python
df = pd.DataFrame({"date":["05/06/2024", "12/01/2024", "25/03/2024"]})
print(df)
print(pd.to_datetime(df["date"] , dayfirst=True)) 
```

```python
df = pd.DataFrame({"date": ["2024-01-15", "not a date", "2024-03-05", None]})

df["clean"] = pd.to_datetime(df["date"], errors="coerce")
print(df)
```

```python
df = pd.DataFrame({"price": [100, 250, 75]})

def add_tax(x):
    return x * 1.18

df["price_with_tax"] = df["price"].apply(add_tax)
print(df)
```

```python
df = pd.DataFrame({"score":[55, 82, 40, 91, 67]})
print(df)

df["grade"] = df["score"].apply(lambda x: "pass" if x>50 else "fail")
print(df)
```

```python
df = pd.DataFrame({"temp":[-5, 0, 15, 30, 42]})
print(df)
def classify(x):
    if x<=0:
        return "Freezing"
    elif x>=1 and x<=25:
        return "Mild"
    else:
        return "Hot"
df["condition"] = df["temp"].apply(classify)
print(df)
```

```python
df = pd.DataFrame({"grade": ["A", "B", "C"]})

grade_map = {"A": 90, "B": 80, "C": 70}
df["score"] = df["grade"].map(grade_map)
print(df)
```

```python
df = pd.DataFrame({"grade": ["A", "B", "Z"]})
df["score"] = df["grade"].map(grade_map)
print(df)
# "Z" isn't in the dictionary -> becomes NaN
```

```python
df = pd.DataFrame({"day":["Mon", "Tue", "Wed", "Sat", "Sun"]})
print(df)
df["full_day"] = df["day"].map({"Mon":"Monday","Tue":"Tuesday","Wed":"Wednesday","Thu":"Thursday","Fri":"Friday","Sat":"Saturday","Sun":"Sunday"})
print(df)
partial_map = {"Mon":"Monday","Tue":"Tuesday","Wed":"Wednesday","Thu":"Thursday","Fri":"Friday"}
df["full_day2"] = df["day"].map(partial_map)
print(df)
```

```python
df = pd.DataFrame({"grade": ["A", "B", "C", "D"]})

df["grade"] = df["grade"].replace({"A": "Excellent", "B": "Good"})
print(df)
df["grade"] = df["grade"].replace({"C":"Average"})
print(df)
```

```python
df = pd.DataFrame({"status":["active", "inactive", "pending", "active", "unknown"]})
print(df)
df["status"] = df["status"].replace({"active":"Active","inactive":"Inactive"})
print(df)
```
