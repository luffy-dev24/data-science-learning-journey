```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Alice", "Bob"],
    "math": [90, 70],
    "science": [85, 75]
})
print(df)
```

```python
long_df = pd.melt(df , id_vars="name", value_vars=["math","science"],var_name="subject",value_name="marks")
print(long_df)
```

```python
df = pd.DataFrame({
    "name":["Alice", "Bob"],
    "jan":[200, 150],
    "feb":[220, 180]
})
print(df)
```

```python
long_df = pd.melt(df,id_vars="name",value_vars=["jan","feb"],var_name="month",value_name="sales")
print(long_df)
```

```python
df = pd.DataFrame({
    "name":["teja","raju"],
    "city":["hyd","che"],
    "q1":[22,33],
    "q2":[21,34],
    "q3":[66,88]
})
print(df)
```

```python
long_df = pd.melt(df , id_vars=["name","city"])
print(long_df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Alice", "Bob"],
    "month": ["Jan", "Jan", "Feb", "Feb"],
    "sales": [200, 150, 220, 180]
})
print(df)
```

```python
wide_df = df.pivot_table( index="name",columns="month", values="sales")
print(wide_df)
```

```python
df =pd.DataFrame({
    "name":["teja","raju","teja","raju"],
    "subjects":["math","english","english","math"],
    "score":[45,55,43,44]
})
print(df)
```

```python
wide_df = pd.pivot_table(df , index="name" , columns="subjects" , values="score")
print(wide_df)
```

```python
df = pd.DataFrame({
    "team":["A","B"],
    "2022":[2,2],
    "2023":[2,1],
    "2024":[1,2]
})
print(df)
```

```python
long_df = pd.melt(df , id_vars="team",value_vars=["2022","2023","2024"],value_name="wins",var_name="year")
print(long_df)
```

```python
wide_df = pd.pivot_table(long_df , index="team" , columns="year" , values="wins")
print(wide_df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "city": ["Hyd", "Hyd", "Che", "Che"],
    "product": ["Pen", "Book", "Pen", "Book"],
    "units_sold": [30, 50, 25, 40]
})
print(df)
```

```python
long_df = pd.pivot_table(df , index="city" , columns="product" , values="units_sold")
print(long_df)
```
