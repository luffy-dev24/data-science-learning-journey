```python
import pandas as pd
```

```python
df_raw = pd.read_csv("gym_bookings_raw.csv")
```

```python
df = df_raw
```

```python
df
```

```python
df["member_name"] =  df["member_name"].str.strip().str.lower()
```

```python
df["branch"] = df["branch"].str.strip().str.lower()
```

```python
df["trainer"] = df["trainer"].str.strip().str.lower()
```

```python
df["class_type"] = df["class_type"].str.strip().str.lower()
```

```python
df["membership_type"] = df["membership_type"].str.strip().str.lower()
```

```python
df["member_rating"] = df["member_rating"].fillna(df["member_rating"].mean())
```

```python
df["booking_date"] = pd.to_datetime(df["booking_date"] , format="mixed" , errors="coerce")
```

```python
df["session_minutes"] = df["session_minutes"].fillna(df["session_minutes"].mean())
```

```python
df["session_minutes"] = df["session_minutes"].astype(int)
```

```python
df = df[ df["session_minutes"] > 0 ]
```

```python
df
```

```python
df = df.dropna()
```

```python
df
```

```python
df = df[ df["session_minutes"] < 180 ]
```

```python
df
```

```python
df.duplicated(df.columns.drop("booking_id"))
```

```python
df = df.drop_duplicates(df.columns.drop("booking_id"))
```

```python
df = df.reset_index(drop=True)
```

```python
df = df[ df["fee"] > 0 ]
```

```python
df
```

```python
#finding average fee per branch
df.groupby("branch")["fee"].mean()
```

```python
#finding average rating per branch and class type
df.groupby(["branch","class_type"])["member_rating"].mean()
```

```python
df.groupby("trainer")["fee"].agg(["mean","max","min"])
```

```python
df["branch_avg_fee"] = df.groupby("branch")["fee"].transform("mean")
```

```python
df["fee_vs_branch_avg"] = df["fee"] - df["branch_avg_fee"]
```

```python
df
```

```python
df_filtered = df.groupby("branch").filter(lambda x: x["member_rating"].mean() > 4.2)
```
