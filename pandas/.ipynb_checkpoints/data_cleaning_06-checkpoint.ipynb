```python
import pandas as pd
```

```python
df_raw = pd.read_csv("streaming_viewership_raw.csv")
```

```python
df = df_raw
```

```python
df.info()
```

```python
df.describe()
```

```python
df
```

```python
df["genre"] = df["genre"].str.strip().str.lower()
```

```python
df["device"] = df["device"].str.strip().str.lower()
```

```python
df["plan"] = df["plan"].str.strip().str.lower()
```

```python
df = df.drop_duplicates(subset=df.columns.drop("view_id"))
```

```python
df
```

```python
df["user_rating"] = df["user_rating"].fillna(0)
```

```python
df["user_age"].fillna(df["user_age"].mean() , inplace = True)
```

```python
df["user_age"] =  df["user_age"].astype(int)
```

```python
import numpy as np
```

```python
df.loc[df["watch_minutes"] < 0, "watch_minutes"] = np.nan
```

```python
df.loc[df["watch_minutes"] > 300, "watch_minutes"] = np.nan 
```

```python
df["watch_minutes"] = df["watch_minutes"].fillna(df["watch_minutes"].mean())
```
