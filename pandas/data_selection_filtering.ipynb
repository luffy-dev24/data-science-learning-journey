```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
```

```python
df
```

```python
#selecting columns
print(df["total_bill"]) #selecting single column 

print(df[["total_bill","tip"]]) #selecting multiple columns
```

```python
#selecting rows
df.loc[0] #selecting a single row
```

```python
df.iloc[0] #selecting a single row by position
```

```python
df.loc[0, "total_bill"]  #selecting a specific row and column by label
```

```python
df.iloc[0 , 0] #selecting a specific row and column by position
```

```python
df.loc[0:2,["total_bill","tip"]] #slicing , the last row in inclusive
```

```python
df.iloc[0:2,0:2]  #slicing , the last row in exclusive
```

```python
#fast accessing a single element
df.at[0,"total_bill"]
```

```python
df.iat[0,0]
```

```python
#filtering and conditions
```

```python
df[df["sex"] == "Male"] #getting only male rows
```

```python
df[df["total_bill"]>30][0:4] #filtering and slicing
```

```python
#multiple conditions AND OR
```

```python
df[ ( df["total_bill"]>30 ) & ( df["sex"]== "Male" ) ]
```

```python
df[ (df["total_bill"] > 45 ) | ( df["sex"] == "Male" ) ]
```

```python
#querying with .query() method 
```

```python
df.query(" total_bill > 30 and time == 'Dinner' ")
```

```python
df.query("day == 'Sun' ")
```

```python
col = "total_bill"
df.query(f"{col} > 25")
```

```python
col = "day"
df.query(f"{col} == 'Sun' ")
```

```python
totalbill = 30
df.query("total_bill > @totalbill")
```
