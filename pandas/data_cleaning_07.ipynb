```python
import pandas as pd
```

```python
df_raw = pd.read_csv("food_delivery_raw.csv")
```

```python
df = df_raw
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df["restaurant"] = df["restaurant"].str.strip().str.lower()
```

```python
df["cuisine"] = df["cuisine"].str.strip().str.lower()
```

```python
df["payment_mode"] = df["payment_mode"].str.strip().str.lower()
```

```python
df["customer_city"] = df["customer_city"].str.strip().str.lower()
```

```python
df
```

```python
df = df.drop_duplicates(df.columns.drop(["order_id","delivery_rating"]))
```

```python
df
```

```python
import numpy as np
```

```python
df.loc[df["order_amount"] < 0 , "order_amount"] = np.nan
```

```python
df.loc[ (df["delivery_minutes"] > 120) | (df["delivery_minutes"] < 0 ) , "delivery_minutes" ]= np.nan
```

```python
df["delivery_rating"] = df["delivery_rating"].fillna(0)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"] , format="mixed" , errors="coerce")
```

```python
df["order_date"].isnull()
```

```python
df_deliver_minutes = df["delivery_minutes"]
```

```python
df["delivery_minutes"] = df_deliver_minutes.fillna(df_deliver_minutes.mean())
```

```python
df["delivery_minutes"] = df["delivery_minutes"].astype(int)
```

```python
df
```

```python
df.dropna(inplace=True)
```

```python
df.reset_index(drop=True , inplace=True)
```
