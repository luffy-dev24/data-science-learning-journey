```python
import numpy as np
```

```python
stock_prices = "1420, 1385, 1402, 1450, 1398, 1411, 1467" #closing stock prices task 1
```

```python
new_stock_price_list = list(map(lambda x:int(x),stock_prices.split(",")))
new_stock_price_list
```

```python
#creating numpy array from these prices
stock_prices_array = np.array(new_stock_price_list)
stock_prices_array
```

```python
#properties of this array
```

```python
print(f"shape of array : {stock_prices_array.shape} ")
print(f"size of array : {stock_prices_array.size}")
print(f"dimensions of array : {stock_prices_array.ndim}")
print(f"data type of array : {stock_prices_array.dtype}")
```

```python
#changing the datatype
stock_prices_array.astype(np.float32)
```
