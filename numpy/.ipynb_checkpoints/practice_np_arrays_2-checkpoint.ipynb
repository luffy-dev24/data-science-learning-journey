```python
import numpy as np

#task 1 attendence percentages of 10 days
```

```python
attendence_percentages = np.array([78, 82, 90, 65, 70, 88, 95, 60, 73, 85])
#print the first and last percentages

print(attendence_percentages[0])
print(attendence_percentages[-1])
```

```python
#slice out the days from 3 to 7
print(attendence_percentages[2:7])
```

```python
#Print attendance on every alternate day using step slicing
print(attendence_percentages[::2])
```

```python
#Reverse the entire array
print(attendence_percentages[::-1])
```

```python
#task 2 view trap
sales = np.array([200, 250, 300, 150, 400])
sales
```

```python
subset = sales[0:3]
print(subset)
subset[0] = 0
print(subset)
print(sales)
#in numoy slicing does not give a new array , it gives a view from the orginal array 
```

```python
#task 3 
orders = np.array([450, 1200, 300, 5000, 800, 150, 2200, 90])
orders
```

```python
#Use boolean masking to get all orders above ₹1000
mask = orders > 1000
print(f"orders greater than 1000 : {mask}")
highest_value=orders[mask]
print(f"high value orders : {highest_value}")
```

```python
#Use fancy indexing to fetch orders at positions [1, 3, 6] directly.
idx = [1,3,6]
new_orders = orders[idx]
print(new_orders)
```

```python
#Replace all orders below ₹200 with 0
orders[orders<200] = 0
print(f"orders after zeroing the below 200 to 0 : {orders}")
```

```python
#task 4 Classroom marks matrix
```

```python
student_marks = np.array([[85, 90, 78], [70, 88, 95], [60, 82, 75], [91, 68, 73]])
student_marks
```

```python
#Get the marks of the 2nd student (all subjects) using row indexing.
print(student_marks[1])
```

```python
#Get everyone's Science marks (column 1) using column slicing.
print(student_marks[: , 1])
```

```python
# Find the total marks per student using axis=1
print(f"total marks per student : { np.sum(student_marks,axis=1) }")
```

```python
#Find the average marks per subject using axis=0 with np.mean
print(f"average marks per subject : { np.mean(student_marks , axis=0) }")
```

```python
#Set everyone's English marks (column 2) to 0 
student_marks[ : , 2] = 0
student_marks
```
