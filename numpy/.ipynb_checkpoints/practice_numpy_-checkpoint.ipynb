```python
import numpy as np
#creating arrays
arr = np.array([1,2,3,4,5])
print(f" 1D Array : {arr}")

arr2 = np.array([ [1,2,3],[4,5,6]])
print(f" 2D array : {arr2}")
```

```python
#creating array from scratch

arr_zero = np.zeros((3, 3))
print(f"zero arrays : {arr_zero}")
print()
arr = np.zeros((2,3))
print(arr)
print()
arr = np.zeros((3,2))
print(arr)
```

```python
arr = np.ones((2,3))
print(arr)
print()
arr = np.ones((3,4))
print(arr)
```

```python
arr = np.full((2,3),6)
print(arr)
print()
arr = np.full((3,3),5)
print(arr)
```

```python
arr = np.eye(4)
print(arr)
```

```python
arr = np.arange(1,20,2)
print(arr)
```

```python
arr = np.linspace(0, 1, 3)
print(arr)
```

```python
arr = np.array([[10,20,30],[40,50,60]])
print(arr)
print(f"dimensions : {arr.ndim}") # 2D array
print(f"shape : {arr.shape}") # 2,3
print(f" size : {arr.size}") # elements in the array - 6
print(f"data type : {arr.dtype}") # int 64
```

```python
arr = np.array([1, 2, 3], dtype=np.float32)  # Explicit type
print(arr.dtype)  # float32

arr_int = arr.astype(np.int32)  # Convert float to int
print(arr_int)  # [1 2 3]
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)  # (2, 3)

print()
reshaped = arr.reshape((3, 2))  # Change shape
print(reshaped)
# [[1 2]
#  [3 4]
#  [5 6]]

print()
flattened = arr.flatten()  # Convert 2D → 1D
print(flattened)  # [1 2 3 4 5 6]
```

```python
# Simulating a dataset (5 samples, 3 features)
data = np.array([[10, 20, 30],
                 [15, 25, 35],
                 [20, 30, 40],
                 [25, 35, 45],
                 [30, 40, 50]])

# Calculating mean and standard deviation for each feature (column)
mean = data.mean(axis=0)
print(mean)
std = data.std(axis=0)
print(std)

# Normalizing the data using broadcasting
normalized_data = (data - mean) / std

print(normalized_data)
```
