```python
#task 1 , Find the average
scores = np.array([45, 60, 78, 90, 55])
print("Scores:", scores)
```

```python
average_score = np.mean(scores)
print("Average score:", average_score)
```

```python
#task 2 , Handle a missing score
scores_with_missing = np.array([45, 60, 78, 90, 55, np.nan])
scores_with_missing
```

```python
# nanmean = "give me the mean, but ignore missing values"

avg_ignoring_missing = np.nanmean(scores_with_missing)
print("Average ignoring missing:", avg_ignoring_missing)

# Now fill the missing spot with that average
scores_with_missing[np.isnan(scores_with_missing)] = avg_ignoring_missing
print("Fixed scores:", scores_with_missing)
```

```python
#task 3 , Check "how wrong" a prediction was
actual =    np.array([45, 60, 78, 90, 55])
predicted = np.array([50, 58, 80, 85, 60])
```

```python
errors = actual - predicted
print("Errors:", errors)
```

```python
average_error = np.mean(np.abs(errors))  # np.abs makes negative errors positive
print("Average error:", average_error)
```
