Simple Event :
  A simple event is when you're looking at just one single outcome — nothing combined, nothing extra.

```python
#Example: Rolling a die and getting exactly a "3." That's it — one specific outcome, nothing more.

import numpy as np

six_sided_die = [1,2,3,4,5,6]

picks = np.random.choice(six_sided_die , size=10000)

count_of_3 = np.sum(picks == 3)

simulated_probability = count_of_3 / len(picks)

print(f"Probability of getting exactly 3 : {simulated_probability}")

```

Compound Event :
    A compound event is when you're looking at more than one outcome combined — basically, a simple event "grouped up."

```python
#Example: Rolling a die and getting an even number (2, 4, or 6). 
#You're not asking about just one outcome — you're combining three of them into one event.


six_sided_die = [1,2,3,4,5,6]

picks = np.random.choice(six_sided_die , size=10000)

count_of_even_numbers = np.sum(picks%2 == 0)

simulated_probability = count_of_even_numbers / len(picks)

print(f"Probability of getting even numbers : {simulated_probability}")
```

Mutually Exclusive Events :
    Two events are mutually exclusive if they cannot both happen at the same time. If one happens, the other definitely can't

```python
#Example: Rolling a die and getting a "3" AND getting a "5" in the same roll. 
#Impossible — a single roll only gives one number. So "getting a 3" and "getting a 5" are mutually exclusive.

# Mutually exclusive: P(3) and P(5) can never happen together on ONE roll
die = [1,2,3,4,5,6]
picks = np.random.choice(die, size=10000)

# check how many times we got BOTH 3 and 5 at once (in a single roll)
# since one roll gives one number, this should always be impossible → count = 0
both_3_and_5 = np.sum((picks == 3) & (picks == 5))
print(both_3_and_5)  # will always print 0 — proving they're mutually exclusive
```

Independent Events :
    Two events are independent if the outcome of one does not affect the outcome of the other at all.
    

```python
#Example: Flipping a coin, then rolling a die. 
#Whether you got heads or tails has zero effect on what number the die shows. These two events don't influence each other.

# Independent events: coin flip result doesn't affect die roll result
coin = np.random.choice([1, 2], size=10000)   # 1 = head, 2 = tail
die_roll = np.random.choice([1,2,3,4,5,6], size=10000)

# probability of head AND rolling a 3 (two independent events together)
p_head_and_3 = np.sum((coin == 1) & (die_roll == 3)) / len(coin)
print(p_head_and_3)  # should be close to (1/2) * (1/6) = 0.0833
```

Dependent Events :
    Two events are dependent if the outcome of the first event changes the probability of the second one.

```python
#Example: Picking 2 cards from a deck without putting the first one back. 
#Once you remove the first card, the deck composition changes — 
#so the probability of the second draw depends on what happened in the first draw.

# Dependent events: drawing 2 Aces in a row, WITHOUT replacement
# deck: 4 Aces out of 52 cards

trials = 100000
success = 0

for _ in range(trials):
    deck = [1]*4 + [0]*48   # 1 = Ace, 0 = not Ace
    first_draw = np.random.choice(deck)
    deck.remove(first_draw)   # remove the card we just drew — deck shrinks!
    second_draw = np.random.choice(deck)
    
    if first_draw == 1 and second_draw == 1:
        success += 1

simulated_probability = success / trials
print(simulated_probability)  
# theoretical: (4/52) * (3/51) ≈ 0.00452
```

```python
#practice problem : 
#problem : You roll a die once. Write code to simulate the probability of a compound event: getting a number that is either 1 or 2.

die = [1,2,3,4,5,6]
draws = np.random.choice(die , size=10000)
count_of_1_or_2 = np.sum((draws == 1) | (draws == 2))
simulated_probability = count_of_1_or_2 / len(draws)
print(f"probability of getting either 1 or 2 : {simulated_probability}")
```

```python
#Problem : You flip a coin AND roll a die at the same time. 
#Are these two events independent or dependent? Write code to find the probability of getting Tails AND a 5

coin = np.random.choice([1,2] , size=10000)
die = np.random.choice([1,2,3,4,5,6],size=10000)
#both does not affect each other so these are independent
simulated_probability = np.sum( (coin == 2) & (die == 5))/ len(die)
print(f"probability of getting tail AND 5 {simulated_probability}")
```

```python
#problem : You draw one card from a deck. Are "drawing a King" and "drawing a Queen" mutually exclusive events?
#Explain why in your own words, then write code to prove your answer

deck = [1]*4 + [2]*4 + [0]*44   # 4 Kings + 4 Queens + 44 other cards = 52 total
picks = np.random.choice(deck , size=10000)
count_of_king_and_queen = np.sum( (picks == 1) & (picks == 2) )
simulated_probability = count_of_king_and_queen / len(picks)
print(f"probability of drawing a king and a queen :  {simulated_probability}") #these are mutually exclusive
```

```python
#Problem : You draw 2 cards from a deck without replacement. Is this dependent or independent? 
#Write code to simulate the probability of drawing 2 red cards in a row.

trials = 100000
success = 0

for _ in range(trials):
    deck = [1]*26 + [0]*26   # 1 = red, 0 = black
    first_draw = np.random.choice(deck)
    deck.remove(first_draw)   # remove the card we just drew — deck shrinks!
    second_draw = np.random.choice(deck)
    
    if first_draw == 1 and second_draw == 1:
        success += 1

simulated_probability = success / trials
print(simulated_probability)  #it is dependent 
```
