Probability just tells you how likely something is to happen. It's always a number between 0 and 1.

0 means "will never happen"
1 means "will definitely happen"
0.5 means "50-50 chance"

Formula : 
Probability = (Number of ways the thing you want can happen) / (Total number of possible outcomes)


```python
# problem 1
import numpy as np

# Simulate rolling a die 10,000 times
rolls = np.random.randint(1, 7, size=10000)  # generates numbers between 1 and 6

# Count how many times we got a 4
count_of_4 = np.sum(rolls == 4)

# Calculate probability from simulation
simulated_probability = count_of_4 / len(rolls)

print(f"Simulated probability of rolling a 4: {simulated_probability}")
print(f"Theoretical probability (1/6): {1/6}")
```

```python
# problem 2 : You flip a fair coin once. What is the probability of getting "Heads"?

flip_coin = np.random.randint(1,3,size=10000)

# 1 - head
# 2 - tail

count_of_head = np.sum(flip_coin == 1)

simulated_probability = count_of_head/len(flip_coin)
print(simulated_probability)
```

```python
#Problem 3: A jar has 5 red balls and 3 blue balls (8 total). If you pick one ball randomly, what's the probability it's blue?

# jar: 5 red balls, 3 blue balls
# represent red as 1, blue as 2
jar = [1]*5 + [2]*3   # this creates a list like [1,1,1,1,1,2,2,2]

# now randomly pick from this jar, many times, WITH replacement
picks = np.random.choice(jar, size=10000)

# count how many times we picked blue (2)
count_blue = np.sum(picks == 2)

simulated_probability = count_blue / len(picks)
print(simulated_probability)
```

```python
# problem 4 :  A jar has 3 red balls and 2 blue balls and 5 green balls (10 total). 
#If you pick one ball randomly, what's the probability it's red?

#jar with red,blue,green
# 1 - red
#2 - blue
# 3 - green
jar = [1]*3 + [2]*2 + [3]*5

picks  = np.random.choice(jar , size =10000)

count_of_red = np.sum(picks == 1)

simulated_probability = count_of_red / len(picks)
print(simulated_probability)
```

```python
# Problem 4: You roll a 6-sided die once. 
#What is the probability of getting a number greater than 4?

six_sided_die = [1,2,3,4,5,6]

picks = np.random.choice(six_sided_die ,size=10000)

count_greaterthan_four = np.sum(picks>4)

simulated_probability = count_greaterthan_four / len(picks)
print(simulated_probability)
```

```python
#Problem 5: A deck of cards has 52 cards. What is the probability of drawing an Ace (there are 4 Aces in a deck)?

# deck: 4 Aces, 48 non-Aces
# represent Ace as 1, non-Ace as 0
deck = [1]*4 + [0]*48   # 4 ones, 48 zeros = 52 total

# randomly draw from this deck, many times, WITH replacement
draws = np.random.choice(deck, size=10000)

# count how many times we drew an Ace (1)
count_ace = np.sum(draws == 1)

simulated_probability = count_ace / len(draws)
print(simulated_probability)
```

```python
#Problem: A deck of 52 cards has 26 red cards (hearts + diamonds) and 26 black cards (clubs + spades).
#If you draw one card randomly, what is the probability that it is a red card?

# 1 - red
# 2 - black
cards = [1]*26 + [2]*26 

picks = np.random.choice(cards , size=10000)
count_of_red_cards = np.sum(picks == 1)

simulated_probability = count_of_red_cards / len(picks)
print(simulated_probability)
```
