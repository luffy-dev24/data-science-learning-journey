```python
# practicing multiplication rule of probability

import numpy as np
```

The multiplication rule tells you how to find the probability of "event A AND event B both happening."

Example: Flipping a coin and rolling a die. P(Heads AND rolling a 6)

P(Heads and 6) = (1/2) * (1/6) = 1/12 ≈ 0.0833

```python
#probelm : You flip a coin twice. What is P(getting Heads on both flips)

flip1 = np.random.choice([1, 2], size=10000)   # 1 = head, 2 = tail
flip2 = np.random.choice([1, 2], size=10000)   # a SEPARATE flip


count_both_heads = np.sum((flip1 == 1) & (flip2 == 1))

simulated_probability = count_both_heads / len(flip1)
print(simulated_probability)
```

```python
#Problem 2: A bag has 4 white balls and 6 black balls (10 total).
#You draw 2 balls without replacement. What is P(both balls are black)?

trails = 10000
success =0
for _ in range(trails):
    balls = [1]*4 + [2]*6 # 1- white balls , 2 - black balls
    first_pick = np.random.choice(balls)
    balls.remove(first_pick)
    second_pick = np.random.choice(balls)

    if first_pick == 2 and second_pick == 2:
        success+=1
simulated_probability = success / trails
print(simulated_probability)
```

```python
# Problem 3: You roll a die twice. What is P(getting a 6 on the first roll AND an even number on the second roll)

draw1 = np.random.choice([1,2,3,4,5,6],size=10000)
draw2 = np.random.choice([1,2,3,4,5,6],size=10000)

count_of_6 = np.sum((draw1 == 6) & (draw2 % 2 == 0))

simulated_probability = count_of_6 / len(draw1)
print(simulated_probability)
```

```python
# problem 4: A deck of 52 cards. You draw 2 cards without replacement. 
#What is P(first card is a King AND second card is a Queen)?

trails = 10000
success = 0

for _ in range(trails):
    cards = [0]*44 + [1]*4 + [2]*4  # 1-king , 2- queen
    first_pick = np.random.choice(cards)
    cards.remove(first_pick)
    second_pick = np.random.choice(cards)

    if first_pick == 1 and second_pick == 2:
        success+=1
simulated_probability = success / trails
print(simulated_probability)
```

```python
#Problem 5: A box has 3 defective bulbs and 7 working bulbs (10 total). 
#You pick 2 bulbs without replacement. What is P(both bulbs are defective)?

trails = 10000
success =0

for _ in range(trails):
    bulbs = [1]*3 + [2]*7 # 1-defective bulbs , 2- working bulbs
    first_pick = np.random.choice(bulbs)
    bulbs.remove(first_pick)
    second_pick = np.random.choice(bulbs)

    if first_pick == 1 and second_pick == 1:
        success +=1
simulated_probability = success/trails
print(simulated_probability)
```
