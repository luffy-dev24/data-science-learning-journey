```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#line plot in seaborn
days = [1, 2, 3, 4, 5, 6, 7]
temp = [22, 24, 23, 26, 28, 27, 25]

sns.lineplot(x=days , y=temp , color="red")
plt.title("temperature over a week")
plt.xlabel("days")
plt.ylabel("temperature in (°C)")
plt.show()
```

```python
#line plot of students marks 
tests = [1, 2, 3, 4, 5]
marks = [60, 65, 70, 68, 75]
marks_2 = [55, 60, 58, 65, 70]

sns.set_theme(style="darkgrid")
sns.lineplot(x=tests ,y=marks,color="orange" ,label="std-1")
sns.lineplot(x=tests ,y=marks_2 , color="red" ,label="std-2")
plt.title("students marks over five tests")
plt.xlabel("tests")
plt.ylabel("marks scored")
plt.xticks(tests)
plt.show()
```

```python
# bar chart in seaborn

subjects = ["Math", "Science", "English", "History"]
scores = [85, 90, 78, 88]

sns.barplot(x=subjects ,y=scores)
plt.title("comparing marks of students in subjects")
plt.xlabel("subjects")
plt.ylabel("marks scored")
plt.show()
```

```python
flavors = ["Vanilla", "Chocolate", "Strawberry", "Mango"]
sales = [120, 150, 90, 110]

sns.barplot(x=flavors,y=sales)
plt.title("sales of icecreams")
plt.xlabel("flavors")
plt.ylabel("sales")
plt.show()
```

```python
import pandas as pd

tests = [1, 2, 3, 4, 5] * 2  # repeated for both students
marks = [60, 65, 70, 68, 75] + [55, 60, 58, 65, 70]
student = ["Student 1"]*5 + ["Student 2"]*5

df = pd.DataFrame({"test": tests, "marks": marks, "student": student})

sns.barplot(x="test", y="marks", hue="student", data=df)
plt.title("Marks Comparison per Test")
plt.xlabel("Test")
plt.ylabel("Marks")
plt.show()
```

```python
# scatter plots
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 70, 78, 85]

sns.set_theme(style="whitegrid")
sns.scatterplot(x=study_hours ,y=marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

for i in range(len(study_hours)):
    plt.annotate(f"st-{i+1}" , (study_hours[i]+0.1 , marks[i]))
plt.show()
```

```python
ad_spend = [10, 20, 30, 40, 50, 60]
sales = [15, 35, 40, 50, 65, 80]

sns.set_theme(style="darkgrid")
sns.scatterplot(x=ad_spend , y=sales)
plt.title("AD spend VS Sales")
plt.xlabel("AD spend")
plt.ylabel("Sales")
plt.show()
```
