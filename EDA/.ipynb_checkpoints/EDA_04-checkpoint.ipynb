```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Invoice_ID": range(1001, 1021),

    "Branch": ["A","A","B","C","A","B","C","A","B","C",
               "A","B","C","A","B","C","A","B","C","A"],

    "City": ["Hyderabad","Hyderabad","Bangalore","Chennai",
             "Hyderabad","Bangalore","Chennai","Hyderabad",
             "Bangalore","Chennai","Hyderabad","Bangalore",
             "Chennai","Hyderabad","Bangalore","Chennai",
             "Hyderabad","Bangalore","Chennai","Hyderabad"],

    "Customer_Type": ["Member","Normal","Member","Member",
                      "Normal","Member","Normal","Member",
                      "Normal","Member","Normal","Member",
                      "Normal","Member","Member","Normal",
                      "Member","Normal","Member","Normal"],

    "Gender": ["Male","Female","Female","Male","Female",
               "Male","Female","Male","Male","Female",
               "Female","Male","Female","Male","Female",
               "Male","Female","Male","Female","Male"],

    "Product_Line": ["Electronics","Food","Clothing","Food",
                     "Electronics","Sports","Clothing",
                     "Sports","Food","Electronics",
                     "Food","Sports","Electronics",
                     "Clothing","Sports","Food",
                     "Electronics","Clothing","Sports","Food"],

    "Unit_Price": [25000,200,1500,180,22000,3500,1800,
                   4000,220,26000,250,3700,24000,
                   1700,3600,210,23000,1600,3900,230],

    "Quantity": [1,5,2,8,1,2,3,1,6,1,
                 4,2,1,2,2,7,1,3,2,5],

    "Rating": [9.2,8.5,8.9,7.8,9.5,8.8,8.4,9.1,
               np.nan,9.7,8.0,8.9,9.6,8.3,8.7,
               7.9,9.4,8.6,8.8,8.1]
}

df_raw = pd.DataFrame(data)
```

```python
df = df_raw.copy()
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull()
```

```python
df.loc[df["Rating"].isnull() == True , "Rating"] = df["Rating"].mean()
```

```python
df
```

```python
df["Revenue"] = df["Unit_Price"]*df["Quantity"]
```

```python
df
```

```python
def is_revenue(r):
    if r < 5000:
        return "Small"
    elif r>5000 and r<20000:
        return "Medium"
    else:
        return "Large"
df["Revenue_Group"]= df["Revenue"].apply(is_revenue)
```

```python
df
```

```python
#Analysis
```

```python
total_revenue_by_branch = df.groupby("Branch")["Revenue"].sum()
```

```python
total_revenue_by_branch
```

```python
plt.bar(total_revenue_by_branch.index,total_revenue_by_branch.values,color="#8027F5")
plt.xlabel("Branch")
plt.ylabel("Total Revenue")

x = np.array(total_revenue_by_branch.index)
for i in range(len(x)):
    plt.text(i,total_revenue_by_branch.values[i] , str(total_revenue_by_branch.values[i]) , ha="center")
plt.show()

#the branch "A" is genrated highest total revenue
```

```python
df
```

```python
total_revenue_by_products = df.groupby("Product_Line")["Revenue"].sum()
```

```python
total_revenue_by_products
```

```python
plt.bar(total_revenue_by_products.index , total_revenue_by_products.values , color="orange")
plt.xlabel("Products")
plt.ylabel("Total Revenue")
plt.show()
```

```python
labels= [i for i in total_revenue_by_products.index]
print(labels)
products_revenues = [i for i in total_revenue_by_products.values]
colors = ["#8027F5","#46F527","#F58E27","#27A3F5"]
explode = [0,0.1,0,0]
plt.pie(products_revenues,labels=labels , colors=colors , autopct="%1.1f%%" ,
        wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'} , explode=explode)
plt.show()
```

```python
df
```

```python
total_revenue_by_cities = df.groupby("City")["Revenue"].sum()
```

```python
total_revenue_by_cities
```

```python
plt.bar(total_revenue_by_cities.index , total_revenue_by_cities.values , color="skyblue")
plt.xlabel("cities")
plt.ylabel("Total Revenue")
plt.show()
```

```python
cities_label = [i for i in total_revenue_by_cities.index]
print(cities_label)
cities_rev = [i for i in total_revenue_by_cities.values]
print(cities_rev)

explode = [0,0,0.1]

plt.pie(cities_rev ,
        labels=cities_label ,
       autopct = "%1.2f%%",
       wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'} ,
       explode=explode)
plt.show()
```

```python
df["Rating"] = df["Rating"].astype(float).round(1).copy()
```

```python
avg_rating_products = df.groupby("Product_Line")["Rating"].mean()
```

```python
plt.bar(avg_rating_products.index , avg_rating_products.values , color="skyblue")
plt.xlabel("products")
plt.ylabel("average rating")
plt.title("Average Rating Of Products")
plt.show()
```

```python
avg_rating_branch = df.groupby("Branch")["Rating"].mean()
```

```python
plt.bar(avg_rating_branch.index , avg_rating_branch.values , color="lightgreen")
plt.xlabel("Branch")
plt.ylabel("average rating mean")
plt.title("Average Rating of Branch")
plt.show()
```
