```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_raw = pd.read_csv("1_missing_values.csv")
```

```python
df = df_raw.copy()
```

```python
df
```

```python
df["score"] = df["score"].fillna(df["score"].mean())
```

```python
df["attendance"] = df["attendance"].fillna(df["attendance"].mean())
```

```python
df
```

```python
total_avg_by_subjects = df.groupby("subject")["score"].mean()
```

```python
x = np.arange(len(total_avg_by_subjects))

sns.barplot(x=total_avg_by_subjects.index , y=total_avg_by_subjects.values , color="orange")
plt.xlabel("subjects")
plt.ylabel("average score")
plt.title("average score comparision by subjects")

for i in range(len(total_avg_by_subjects)):
    plt.text(x[i] , total_avg_by_subjects.values[i]+0.3 , str(int(total_avg_by_subjects.values[i])))
    
plt.show()
```
