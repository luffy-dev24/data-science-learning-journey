```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Movie": [
        "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
        "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
        "Movie K", "Movie L", "Movie M", "Movie N", "Movie O"
    ],

    "Genre": [
        "Action", "Comedy", "Drama", "Action", "Comedy",
        "Drama", "Action", "Thriller", "Comedy", "Drama",
        "Thriller", "Action", "Comedy", "Drama", "Thriller"
    ],

    "Platform": [
        "Netflix", "Prime", "Netflix", "Disney+", "Prime",
        "Netflix", "Prime", "Netflix", "Disney+", "Prime",
        "Netflix", "Disney+", "Prime", "Netflix", "Disney+"
    ],

    "Views_Millions": [
        12.5, 8.2, 10.1, 15.8, 7.5,
        11.3, 18.2, 9.8, 6.4, 13.5,
        14.7, 20.1, 8.9, np.nan, 11.8
    ],

    "Rating": [
        8.5, 7.2, 8.8, 9.1, 6.9,
        8.3, 9.4, 7.8, 6.5, 8.7,
        8.9, 9.5, 7.4, 8.6, np.nan
    ],

    "Runtime_Minutes": [
        140, 110, 125, 150, 105,
        130, 155, 120, 100, 135,
        145, 160, 115, 128, 138
    ]
}

df_raw = pd.DataFrame(data)
```

```python
df = df_raw
df
```

```python
df.describe()
```

```python
df["Views_Millions"] = df["Views_Millions"].fillna(df["Views_Millions"].median()).copy()
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].mean()).copy()
```

```python
df["Rating"] = df["Rating"].astype(float).round().copy()
```

```python
df
```

```python
max_views_by_genre = df.groupby("Genre")["Views_Millions"].max()
```

```python
max_views_by_genre
```

```python
# max millions views of genre


plt.bar(max_views_by_genre.index , max_views_by_genre.values , color="skyblue")
plt.xlabel("Genre")
plt.ylabel("Max Views in Millions")
plt.title("Maximum Million Views of Each Genre")

for i in range(len(max_views_by_genre)):
    plt.text(i , max_views_by_genre.values[i]+0.3 , str(max_views_by_genre.values[i]) , ha="center")
plt.show()
```

```python
len(max_views_by_genre)
```

```python
#total millions views of each genre

total_views_of_genre = df.groupby("Genre")["Views_Millions"].sum()

plt.bar(total_views_of_genre.index , total_views_of_genre.values , color="orange")
plt.title("Total Views of Each Genre in Millions")
plt.xlabel("Genre")
plt.ylabel("Views in Millions")

for i in range(len(total_views_of_genre)):
    plt.text(i , total_views_of_genre.values[i] , str(total_views_of_genre.values[i]) , ha="center")
plt.show()
```

```python
max_ratings_by_genre = df.groupby("Genre")["Rating"].max()
```

```python
# max Rating of each genre

plt.bar(max_ratings_by_genre.index , max_ratings_by_genre.values , color="skyblue")
plt.title("maximum ratings of each genre")
plt.xlabel("Genre")
plt.ylabel("maximum ratings")
plt.show()
```

```python
# total runtimes of each genre

total_runtime_genre = df.groupby("Genre")["Runtime_Minutes"].sum()

plt.bar(total_runtime_genre.index , total_runtime_genre.values , color="orange")
plt.title("runtime of each genre in minutes")
plt.xlabel("Genre")
plt.ylabel("runtime minutes")
plt.show()
```

```python
# total views of each platform 
total_views_platform  = df.groupby("Platform")["Views_Millions"].sum()
total_views_platform

plt.bar(total_views_platform.index , total_views_platform.values , color="skyblue")
plt.title("total views in millions of each platform")
plt.xlabel("Platforms")
plt.ylabel("total views in millions")

for i in range(len(total_views_platform)):
    plt.text(i , total_views_platform.values[i]+0.3 , str(total_views_platform.values[i]) , ha="center")
plt.show()
```

```python
netflix_total_runtime = df[df["Platform"] == "Netflix" ].copy()
netflix_total_runtime_by_genre = netflix_total_runtime.groupby("Genre")["Runtime_Minutes"].sum()

labels =  [i for i in netflix_total_runtime_by_genre.index]
values = [i for i in netflix_total_runtime_by_genre.values]
colors = ["orange","skyblue","lightgreen"]
explode = [0,0.1,0]

plt.pie(values , labels=labels,colors=colors , explode=explode,autopct="%1.1f%%")
plt.show()
```
